```python
def prime(num):
  if num==0 or num==1:
    return 0

  for i in range(2,num):
    if num % i ==0:
      return 0


    return 1

```

```python
count = 1

for i in range (100):
  if prime(i)==1:
    print(count , "\t",i)
    if(count>=20):
      break
    count +=1


```
