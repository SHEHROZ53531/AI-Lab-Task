Creation of a Set:

```python
# Creating a set
pset = {1, 2, 3, 4, 5}
print(pset)
print("Type of pset is :", type(pset))

# Creating a set using constructor
pset1 = set([3, 4, 5, 6, 7])
print(pset1)
print("Type of pset is :", type(pset1))
```

Adding and Removing Set Elements:

```python
pset2 = {5, 20, 11, 21, 8, 9}
print(pset2)

pset2.add(10)
print("pset after adding item value 10: ", pset2)
pset2.remove(20)
pset2.discard(8)
pset2.discard(8)

print(pset2)
```

Set Operations:

```python
# Union of two sets
print("pset1", pset1)
print("pset2", pset2)
pset2.remove(5)
union_set = pset1.union(pset2)
print("Union set after union :", union_set)

# Intersection of two sets
pset1.add(10)
pset3 = pset1.intersection(pset2)
print("Result of intersection of two sets : pset1 and pset2", pset3)

# Difference of two sets
pset4 = union_set.difference(pset2)
print("Difference of two sets : union_set and pset2", pset4)
```

Python Dictionaries

```python
# Defining the dictionary
dict1 = {
    'Name': 'Ahmed',
    'Age': 12,
    'Class': 'Five'
}

print("Dictionary No 1: ", dict1)
print("Type of dict1 is :", type(dict1))

# Accessing Items using square brackets
print("dict1['Name']: ", dict1['Name'])
print("dict1['Age']: ", dict1['Age'])

# Accessing items using the .get() method
x = dict1.get('Age')
print("Getting item through get method", x)

# Getting all keys using the .keys() method
y = dict1.keys()
print("Getting keys through get method", y)
```

Update Dictionary

```python
# 1. Initialize the dictionary
student = {
    'Name': 'Ahmed',
    'Age': 12,
    'Class': 'Five'
}
print(f"Original: {student}")

# 2. Update via Direct Assignment
# Best for changing a single known key
student['Age'] = 15
print(f"Direct Assignment: {student}")

# 3. Update via .update() method
# Best for updating multiple values at once or merging dictionaries
student.update({"Age": 16, "City": "Cairo"})
print(f"Using .update(): {student}")

# 4. Update via Merge Operator (Python 3.9+)
# This creates a new dictionary without modifying the original 'student'
new_data = {"Age": 17, "Grade": "A"}
student = student | new_data
print(f"Using Merge (|): {student}")
```

Removing Items

```python
# Initializing the dictionary
dict1 = {
    'Name': 'Ahmed',
    'Age': 12,
    'Class': 'Five'
}

# Displaying the original dictionary
print("Dictionary No 1: ", dict1)

# Removing the item with the key 'Class' using pop()
dict1.pop('Class')

# Displaying the dictionary after removal
print("Dictionary after pop (class): ", dict1)
```

Removing Dictionary through PopItem

```python
dict1 = {
    'Name': 'Ahmed',
    'Class': 'Five',
    'Age': 12
}

# Removes the last item and stores it
removed_item = dict1.popitem()

print(f"Removed: {removed_item}")
print(f"Updated Dictionary: {dict1}")
```

Deleting Item Through Del

```python
dict1 = {
    'Name': 'Ahmed',
    'Class': 'Five',
    'Age': 12
}

print("Dictionary No 1: ", dict1)

# Deleting the item with the key 'Class'
del dict1['Class']

print("Delete item using del : ", dict1)
```

Removing Dictionary

```python
# Defining the dictionary
dict1 = {
    'Name': 'Ahmed',
    'Class': 'Five',
    'Age': 12
}

# Printing the initial dictionary
print("Dictionary No 1: ", dict1)

# Using clear() to remove all items
dict1.clear()

# Printing the dictionary after clearing it
print("Empty the Dictionary using clear() : ", dict1)
```
