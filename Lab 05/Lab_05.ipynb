```python
from collections import deque

def bfs(start_node, graph, visited, component):
    # Initialize a queue and add the starting node
    queue = deque([start_node])
    visited[start_node] = True  # Mark the start node as visited

    while queue:
        # Dequeue a node from the front of the queue
        node = queue.popleft()
        component.append(node)  # Store answer

        # Traverse to each adjacent node
        for child in graph[node]:
            if not visited[child]:  # Check if visited
                visited[child] = True  # Mark as visited before pushing to queue
                queue.append(child)    # Enqueue the child

if __name__ == "__main__":

    # Graph of nodes
    graph = {
        0: [2],
        1: [2, 3],
        2: [0, 1, 4],
        3: [1, 4],
        4: [2, 3]
    }

    node = 0  # Starting node
    visited = [False] * len(graph)  # Make all nodes False initially
    component = []

    bfs(node, graph, visited, component)  # Traverse using BFS
    print(f"Following is the Breadth-first search: {component}")  # Print the answer
```
